```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 2)

# Create superposition on the first qubit
circuit.h(0)

# Entangle the two qubits
circuit.cx(0, 1)

# Measure both qubits
circuit.measure([0, 1], [0, 1])

print(circuit.draw())

backend = AerSimulator()
simulation = backend.run(circuit, shots=1024).result()

print("Measurement output:", simulation.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def create_bell_state(name):
    circuit = QuantumCircuit(2, 2)

    # Generate the initial Bell state
    circuit.h(0)
    circuit.cx(0, 1)

    # Modify the state according to the selected type
    if name == "Phi-":
        circuit.z(0)
    elif name == "Psi+":
        circuit.x(0)
    elif name == "Psi-":
        circuit.x(0)
        circuit.z(0)

    # Measure both qubits
    circuit.measure([0, 1], [0, 1])
    return circuit


backend = AerSimulator()
bell_states = ["Phi+", "Phi-", "Psi+", "Psi-"]

for name in bell_states:
    circuit = create_bell_state(name)
    output = backend.run(circuit, shots=1024).result().get_counts()

    print("Bell state:", name)
    print("Results:", output)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 2)

# Prepare the Bell state
circuit.h(0)
circuit.cx(0, 1)

# Convert to Hadamard measurement basis
for qubit in range(2):
    circuit.h(qubit)

# Measure both qubits
circuit.measure([0, 1], [0, 1])

print(circuit.draw())

backend = AerSimulator()
simulation = backend.run(circuit, shots=1024).result()

print("Measurement results:", simulation.get_counts())
```
