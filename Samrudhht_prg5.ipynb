```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

print("EXERCISE 1: HADAMARD SUPERPOSITION")
print("=" * 50)

for shots in [100, 1000, 10000]:

    qc = QuantumCircuit(1, 1)

    qc.h(0)
    qc.measure(0, 0)

    result = simulator.run(
        qc,
        shots=shots
    ).result()

    counts = result.get_counts()

    p0 = counts.get("0", 0) / shots
    p1 = counts.get("1", 0) / shots

    print("\nShots:", shots)
    print("Counts:", counts)
    print("P(0):", round(p0, 4))
    print("P(1):", round(p1, 4))
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

shots = 1000

print("EXERCISE 2: X-BASIS MEASUREMENT")
print("=" * 50)


# -------------------------
# |+> STATE
# -------------------------

plus = QuantumCircuit(1, 1)

plus.h(0)      # Create |+>
plus.h(0)      # Change X basis to computational basis
plus.measure(0, 0)

result_plus = simulator.run(
    plus,
    shots=shots
).result()

print("\n|+> STATE")
print(result_plus.get_counts())


# -------------------------
# |-> STATE
# -------------------------

minus = QuantumCircuit(1, 1)

minus.x(0)     # |0> -> |1>
minus.h(0)     # |1> -> |->
minus.h(0)     # X basis -> computational basis
minus.measure(0, 0)

result_minus = simulator.run(
    minus,
    shots=shots
).result()

print("\n|-> STATE")
print(result_minus.get_counts())
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

shots = 1000

print("EXERCISE 3: BELL PAIR PARTIAL MEASUREMENT")
print("=" * 50)

qc = QuantumCircuit(2, 1)

# Create Bell pair
qc.h(0)
qc.cx(0, 1)

# Measure ONLY qubit 0
qc.measure(0, 0)

result = simulator.run(
    qc,
    shots=shots
).result()

counts = result.get_counts()

print("\nMeasurement results:")
print(counts)

print("\nInterpretation:")
print("If qubit 0 = 0 -> qubit 1 collapses to |0>")
print("If qubit 0 = 1 -> qubit 1 collapses to |1>")
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import math

simulator = AerSimulator()

print("EXERCISE 4: UNKNOWN ROTATION ANGLE")
print("=" * 50)

# Actual unknown angle
true_angle = 60

# Convert degrees to radians
theta = math.radians(true_angle)

shots = 10000

qc = QuantumCircuit(1, 1)

# Apply unknown rotation
qc.ry(theta, 0)

# Measure
qc.measure(0, 0)

result = simulator.run(
    qc,
    shots=shots
).result()

counts = result.get_counts()

print("\nMeasurement counts:")
print(counts)

# Probability of measuring 1
p1 = counts.get("1", 0) / shots

print("\nMeasured P(1):", round(p1, 4))

# P(1) = sin²(theta/2)
estimated_theta = 2 * math.asin(
    math.sqrt(p1)
)

estimated_angle = math.degrees(
    estimated_theta
)

print("Actual angle:", true_angle, "degrees")

print(
    "Estimated angle:",
    round(estimated_angle, 2),
    "degrees"
)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from scipy.stats import chisquare
import random

simulator = AerSimulator()

shots = 10000

print("EXERCISE 5: QUANTUM RANDOMNESS TEST")
print("=" * 50)


# ============================================================
# QUANTUM RANDOM GENERATOR
# ============================================================

quantum = QuantumCircuit(1, 1)

quantum.h(0)
quantum.measure(0, 0)

result = simulator.run(
    quantum,
    shots=shots
).result()

counts = result.get_counts()

q0 = counts.get("0", 0)
q1 = counts.get("1", 0)

print("\nQUANTUM GENERATOR")
print("-" * 40)

print("Counts:", counts)

print("P(0):", round(q0 / shots, 4))
print("P(1):", round(q1 / shots, 4))


# ============================================================
# CHI-SQUARE TEST
# ============================================================

expected = [
    shots / 2,
    shots / 2
]

chi_quantum, p_quantum = chisquare(
    [q0, q1],
    f_exp=expected
)

print("\nChi-square:", round(chi_quantum, 4))
print("p-value:", p_quantum)

if p_quantum < 0.05:
    print("Result: Evidence of bias")
else:
    print("Result: Consistent with 50/50 randomness")


# ============================================================
# BIASED CLASSICAL GENERATOR
# ============================================================

classical_0 = 0
classical_1 = 0

for i in range(shots):

    # 70% chance of getting 1
    if random.random() < 0.70:
        classical_1 += 1
    else:
        classical_0 += 1


print("\nBIASED CLASSICAL GENERATOR")
print("-" * 40)

print("0:", classical_0)
print("1:", classical_1)

print("P(0):", round(classical_0 / shots, 4))
print("P(1):", round(classical_1 / shots, 4))


# Chi-square test
chi_classical, p_classical = chisquare(
    [classical_0, classical_1],
    f_exp=expected
)

print("\nChi-square:", round(chi_classical, 4))
print("p-value:", p_classical)

if p_classical < 0.05:
    print("Result: Generator is significantly biased")
else:
    print("Result: No significant evidence of bias")
```
