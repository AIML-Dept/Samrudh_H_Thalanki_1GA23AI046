```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 1)

# Prepare the ancilla qubit
circuit.x(1)

# Create superposition
circuit.h(0)
circuit.h(1)

# Oracle for constant function f(x) = 0
# No operation is needed

# Final Hadamard gate
circuit.h(0)

# Measure the first qubit
circuit.measure(0, 0)

print(circuit.draw())

backend = AerSimulator()
job = backend.run(circuit, shots=1024)
result = job.result()

counts = result.get_counts()
print("Measurement results:", counts)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 1)

# Set the ancilla qubit to |1>
circuit.x(1)

# Apply Hadamard transformation
circuit.h(0)
circuit.h(1)

# Balanced oracle: f(x) = x
circuit.cx(0, 1)

# Apply Hadamard to the first qubit
circuit.h(0)

# Perform measurement
circuit.measure(0, 0)

print(circuit.draw())

backend = AerSimulator()
result = backend.run(circuit, shots=1024).result()

print("Output:", result.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def deutsch_circuit(choice):
    qc = QuantumCircuit(2, 1)

    # Prepare the second qubit
    qc.x(1)

    # Create quantum superposition
    qc.h(0)
    qc.h(1)

    # Apply the selected quantum oracle
    if choice == 0:
        # Constant-zero function
        pass
    elif choice == 1:
        # Constant-one function
        qc.x(1)
    elif choice == 2:
        # Balanced identity function
        qc.cx(0, 1)
    elif choice == 3:
        # Balanced NOT function
        qc.x(1)
        qc.cx(0, 1)
    else:
        raise ValueError("Choice must be between 0 and 3")

    # Measure the input qubit
    qc.h(0)
    qc.measure(0, 0)

    return qc


sim = AerSimulator()
functions = {
    0: "Constant zero",
    1: "Constant one",
    2: "Identity",
    3: "NOT"
}

for key, name in functions.items():
    circuit = deutsch_circuit(key)
    result = sim.run(circuit, shots=1024).result()

    print("\nFunction:", name)
    print("Results:", result.get_counts())
    print(circuit.draw())
```
