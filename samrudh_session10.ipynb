```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 2)

# Prepare a shared Bell pair
circuit.h(0)
circuit.cx(0, 1)

# Alice encodes message 00
# No additional gate is needed

# Bob performs decoding
circuit.cx(0, 1)
circuit.h(0)

# Measure both qubits
circuit.measure([0, 1], [0, 1])

print(circuit.draw())

backend = AerSimulator()
simulation = backend.run(circuit, shots=1024).result()

print("Decoded output:", simulation.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 2)

# Prepare the Bell pair
circuit.h(0)
circuit.cx(0, 1)

# Alice sends message 00
# No encoding operation is needed

# Introduce a bit-flip error
circuit.x(0)

# Bob decodes the message
circuit.cx(0, 1)
circuit.h(0)

# Measure both qubits
circuit.measure([0, 1], [0, 1])

print(circuit.draw())

backend = AerSimulator()
simulation = backend.run(circuit, shots=1024).result()

print("Results after bit-flip:", simulation.get_counts())
```
