```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

num_qubits = 3
circuit = QuantumCircuit(num_qubits + 1, num_qubits)

# Initialize the ancilla qubit
circuit.x(num_qubits)

# Apply Hadamard gates to all qubits
for qubit in range(num_qubits + 1):
    circuit.h(qubit)

# Constant oracle: f(x) = 0
# No additional gate is needed

# Apply Hadamard gates to input qubits
for qubit in range(num_qubits):
    circuit.h(qubit)

# Measure the input register
circuit.measure(range(num_qubits), range(num_qubits))

print(circuit.draw())

backend = AerSimulator()
simulation = backend.run(circuit, shots=1024).result()

print("Measurement results:", simulation.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

num_inputs = 3
circuit = QuantumCircuit(num_inputs + 1, num_inputs)

# Initialize the ancilla qubit
circuit.x(num_inputs)

# Apply Hadamard gates
for qubit in range(num_inputs + 1):
    circuit.h(qubit)

# Balanced parity oracle
for qubit in range(num_inputs):
    circuit.cx(qubit, num_inputs)

# Apply Hadamard gates to input qubits
for qubit in range(num_inputs):
    circuit.h(qubit)

# Measure the input register
circuit.measure(range(num_inputs), range(num_inputs))

print(circuit.draw())

backend = AerSimulator()
simulation = backend.run(circuit, shots=1024).result()

print("Output counts:", simulation.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def run_algorithm(num_inputs, oracle):
    circuit = QuantumCircuit(num_inputs + 1, num_inputs)

    # Prepare the ancilla qubit
    circuit.x(num_inputs)

    # Apply Hadamard gates to all qubits
    for qubit in range(num_inputs + 1):
        circuit.h(qubit)

    # Select the oracle
    if oracle == "constant":
        pass
    elif oracle == "balanced":
        for qubit in range(num_inputs):
            circuit.cx(qubit, num_inputs)
    else:
        raise ValueError("Invalid oracle selection")

    # Apply final Hadamard gates
    for qubit in range(num_inputs):
        circuit.h(qubit)

    # Measure input qubits
    circuit.measure(range(num_inputs), range(num_inputs))
    return circuit


backend = AerSimulator()

for size in range(2, 6):
    for oracle_name in ("constant", "balanced"):
        circuit = run_algorithm(size, oracle_name)
        result = backend.run(circuit, shots=1024).result()

        print("Qubits:", size, "| Oracle:", oracle_name)
        print("Counts:", result.get_counts())
```
